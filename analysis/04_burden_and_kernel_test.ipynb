## Domain based

```python
for i in range(0, 89):
    
    LRRK2_filt_tmp = LRRK2_filt.iloc[i]
    start_pos = LRRK2_filt_tmp["AA_pos"]

    if start_pos in range(1, 12):
        LRRK2_filt.at[i, "DOMAIN"] = "Before_ARM"
        
    elif start_pos in range(12, 705):
        LRRK2_filt.at[i, "DOMAIN"] = "ARM"
        
    elif start_pos in range(705, 796):
        LRRK2_filt.at[i, "DOMAIN"] = "ANK"
        
    elif start_pos in range(796, 800):
        LRRK2_filt.at[i, "DOMAIN"] = "ANK-LRR"
        
    elif start_pos in range(800, 1306):
        LRRK2_filt.at[i, "DOMAIN"] = "LRR"
        
    elif start_pos in range(1306, 1336):    
        LRRK2_filt.at[i, "DOMAIN"] = "LRR-ROC"
        
    elif start_pos in range(1335, 1511):
        LRRK2_filt.at[i, "DOMAIN"] = "ROC"
        
    elif start_pos in range(1511, 1879):
        LRRK2_filt.at[i, "DOMAIN"] = "COR"
        
    elif start_pos in range(1879, 2139):
        LRRK2_filt.at[i, "DOMAIN"] = "KINASE"
        
    elif start_pos in range(2139, 2142):
        LRRK2_filt.at[i, "DOMAIN"] = "KINASE-WD40"
        
    elif start_pos in range(2142, 2498): 
        LRRK2_filt.at[i, "DOMAIN"] = "WD40"
        
    else:
        LRRK2_filt.at[i, "DOMAIN"] = "Outside_domain"

        
```

```python
domains = LRRK2_filt["DOMAIN"].unique()

for domain in domains:
    
    LRRK2_filt_dom = LRRK2_filt[LRRK2_filt["DOMAIN"] == domain]
    
    LRRK2_filt_dom[["ID"]].to_csv(f"{WORK_DIR}/{label}/snp_{domain}_id_wgs.txt", sep = "\t", header = False, index = False)
```

```python
dom = pd.read_csv(f"{WORK_DIR}/{label}/domain_snps_aa_change_wgs.txt", sep = "\t")

domains = dom["Domain"].unique()
domains = domains[::-1]

results = pd.DataFrame()

for domain in domains:
    
    file_path = f"{WORK_DIR}/{label}/GP2_merge_{label}_coding_burden_{domain}_wgs.Skat.assoc"
    
    
    skat = pd.read_csv(f"{file_path}", delim_whitespace = True)
    skat_red = skat[["NumVar", "Pvalue", "PermPvalue"]]
    skat_red["Domain"] = domain

    snps = pd.read_csv(f"{WORK_DIR}/{label}/domain_snps_aa_change_wgs.txt", sep = "\t")
    snps_filt = snps[snps["Domain"] == domain]
    aa_change = ', '.join(snps_filt["AA_Change"].astype(str))

    skat_red["Variants"] = aa_change

    results = pd.concat([skat_red, results])
    
    
results.to_csv(f"{WORK_DIR}/{label}/GP2_merge_{label}_domain_skat_wgs.txt", sep = "\t", header = True, index = False)
```

```python
%%R

label <- "CHINESE"
domains <- c("Before_ARM", "ARM", "ARM-ANK", "ANK", "ANK-LRR", "LRR", "LRR-ROC", "ROC", "ROC-COR", "COR", "KINASE", "KINASE-WD40", "WD40", "Outside_domain")

results <- data.frame()

for (domain in domains) {
    
    file_path <- paste0("/home/jupyter/LRRK2/release11/UMKL/", label, "/GP2_merge_", 
                        label, "_coding_burden_", domain, "_single_wgs.SingleScore.assoc")
    
    if (file.exists(file_path)) {
        
        # Read the single assoc file
        assoc <- read.delim(file_path, sep = "\t")

        # Add domain part
        assoc$DOMAIN <- domain

        # add multiple adjustment p-value
        assoc$FDR_BH <- p.adjust(assoc$PVALUE, method = "BH")
        assoc$FDR_BF <- p.adjust(assoc$PVALUE, method = "bonferroni")

        assoc_red <- assoc %>% select(CHROM, POS, REF, ALT, DOMAIN, AF, 
                                      PVALUE, FDR_BH, FDR_BF, EFFECT, SE, DIRECTION)

        results <- rbind(results, assoc_red)
        
    } else {
        
        print("Skip")
        
    }   
}

results %>% arrange(PVALUE) %>% knitr::kable()
```
