# Haplotype association

```python
%%R

WORK_DIR <- '/home/jupyter/LRRK2/release11/UMKL'
label <- 'CHINESE'

print(paste0("Processing files for ", label))
    
# Read PED file
PED <- fread(paste0(WORK_DIR, "/", label, "/GP2_merge_", label, "_qced_updated_rmrel_LRRK2_nodup.ped"))

# Get Allele name
FILTERED.LRRK2_1 <- read.table(paste0(WORK_DIR, "/", label, "/GP2_merge_", label, "_qced_updated_rmrel_LRRK2_nodup.map"), header=FALSE)
FILTERED.LRRK2_2 <- read.table(paste0(WORK_DIR, "/", label, "/GP2_merge_", label, "_qced_updated_rmrel_LRRK2_nodup.map"), header=FALSE)
FILTERED.LRRK2_1$V2 <- paste0(FILTERED.LRRK2_1$V2, sep="_", "1")
FILTERED.LRRK2_2$V2 <- paste0(FILTERED.LRRK2_2$V2, sep="_", "2")

# Set column names
colnames(FILTERED.LRRK2_1) <- c("CHR", "SNP", "CM", "POS")
colnames(FILTERED.LRRK2_2) <- c("CHR", "SNP", "CM", "POS")
FILTERED.LRRK2_2alleles    <- rbind(FILTERED.LRRK2_1, FILTERED.LRRK2_2)

# Sort file and create geno matrix:
# The variants should appear in order of genomic postition in the ped file, therefore, we sort the alleles based on position
LRRK2_pos     <- FILTERED.LRRK2_2alleles[order(FILTERED.LRRK2_2alleles$POS),]
LRRK2_alleles <- LRRK2_pos[,c("SNP")]
write.table(LRRK2_alleles, file = paste0(WORK_DIR,"/", label, "/", label, "_LRRK2_alleles_nba.txt"), quote = F, sep = "\t", row.names = F, col.names = F)

# Add allele names to the PED file
colnames(PED) <- c("FID", "IID", "PAT","MAT", "SEX", "PHENO", LRRK2_alleles)
write.table(as.data.frame(PED), file = paste0(WORK_DIR,"/", label, "/", label, "_geno_matrix_LRRK2_nba.txt"), quote = F, row.names = FALSE, sep = '\t')
```

```python
%%R

run_haplotype_analysis_adj <- function(label, block) {
    
  # File paths
  base_path <- paste0("/home/jupyter/LRRK2/release11/UMKL/", label, "/")
  geno_matrix_file <- paste0(base_path, label, "_geno_matrix_LRRK2_nba.txt")
  det_file <- paste0(base_path, "/GP2_merge_", label, "_qced_updated_rmrel_LRRK2_nodup_005.blocks.det")
  alleles_file <- paste0(base_path, label, "_LRRK2_alleles_nba.txt")

  # Load data ${label}/cov_glm_${label}_imp.txt
  #Import covariate file and set which variables you would like to adjust for
  SampleInfo_Adjustment <- read.delim(paste0(base_path, "cov_glm_", label, "_raw_hap_nba.txt"), sep="\t") %>%
                           select("SEX","age","PC1","PC2","PC3")
    
  adj <- data.frame(SampleInfo_Adjustment[,c("SEX","age","PC1","PC2","PC3")])
    
  Geno_matrix_LRRK2 <- read.delim(geno_matrix_file, check.names = FALSE)
  det <- read.delim(det_file, sep = "", header = TRUE)
  LRRK2_alleles <- read.table(alleles_file, quote = "\"", comment.char = "")

  # Get SNPs from selected block
  to_grep <- det[block, ]$SNPS

  # Extract relevant SNPs
  H1_a <- LRRK2_alleles[grepl(to_grep, LRRK2_alleles$V1), ]
  H1_LRRK2 <- Geno_matrix_LRRK2[, c("FID", "IID", "PAT", "MAT", "SEX", "PHENO", H1_a)]

  # Extract only genotype columns
  num_col <- ncol(H1_LRRK2)
  geno <- data.frame(H1_LRRK2[, 7:num_col], check.names = FALSE)

  # Set haplotype labels
  haplo_labels <- colnames(H1_LRRK2)[7:num_col] %>%
                    gsub("_.{1}$", "", .) %>%
                    unique()

  # Binary phenotype: 0 = control, 1 = patient
  H1_LRRK2$PHENO_01 <- H1_LRRK2$PHENO - 1
  y.bin <- 1 * (H1_LRRK2$PHENO_01 == 1)

  # Haplotype association test (no covariate)
  H1 <- haplo.cc(
    y = y.bin,
    geno = geno,
    locus.label = haplo_labels,
    x.adj = adj,
    control = haplo.glm.control(haplo.freq.min = 0.01)
  )

  # Sort results by p-value
  H1_cc.df_sort <- H1$cc.df[order(H1$cc.df$`p-val`), ]

  return(list(
    result = H1,
    summary_table = H1_cc.df_sort
  ))
}
```

```python
%%R

# Define ancestry
label <- "CHINESE"
nblock <- 7


# Create empty list to store results
haplo_results <- list()


# Loop through blocks 1 to nblock
for (i in 1:nblock) {
    
  print(paste0("Running block ", i))
    
  result <- tryCatch({
      
    run_haplotype_analysis_adj(label = label, block = i)
      
  }, error = function(e) {
      
    print(paste0("Block ", i, " failed: ", e$message))
    return(NULL)
      
  })
  
  haplo_results[[paste0("Block_", i)]] <- result
}

# Create a new workbook
wb <- createWorkbook()

# Loop through each block in the list and add a worksheet
for (i in 1:nblock) {
  block_name <- paste0("Block_", i)
  result <- haplo_results[[block_name]]
  
  if (!is.null(result)) {
    addWorksheet(wb, sheetName = block_name)
    writeData(wb, sheet = block_name, result$summary_table)
  }
}

base_path <- paste0("/home/jupyter/LRRK2/release11/UMKL/", label, "/")
# Save the workbook
saveWorkbook(wb, file = paste0(base_path, label, "_haplotype_results_r11_nba_005.xlsx"), overwrite = TRUE)

```
